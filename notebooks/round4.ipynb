```python
# pip install -U ultralytics

import os
import torch
from ultralytics import YOLO
```

```python
DATA_YAML = r"D:\MAGFiLO\data.yaml"   # <-- путь к твоему data.yaml
PROJECT_DIR = r"D:\MAGFiLO\runs"      # <-- куда сохранять результаты
MODEL_NAME = "yolo11l.pt"

assert torch.cuda.is_available(), "CUDA/GPU не найдена"
print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)
print("CUDA:", torch.version.cuda)

os.makedirs(PROJECT_DIR, exist_ok=True)

model = YOLO(MODEL_NAME)

results = model.train(
    data=DATA_YAML,
    imgsz=1280,
    epochs=100,
    batch=2,
    device=0,
    workers=4,
    pretrained=True,
    optimizer="AdamW",
    lr0=3e-4,
    lrf=0.01,
    weight_decay=1e-4,
    patience=20,
    cos_lr=True,
    close_mosaic=10,
    amp=True,
    project=PROJECT_DIR,
    name="yolo11l_filament",
    exist_ok=True,
    save=True,
    plots=True,
    verbose=True
)

BEST_PT = os.path.join(PROJECT_DIR, "yolo11l_filament", "weights", "best.pt")

print("\nTraining finished")
print("Best model:", BEST_PT)
print("Exists:", os.path.exists(BEST_PT))
```
